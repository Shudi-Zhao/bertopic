# [Dimensionality Reduction in BERTopic](https://maartengr.github.io/BERTopic/getting_started/dim_reduction/dim_reduction.html)

In BERTopic, **dimensionality reduction** is a crucial step that transforms high-dimensional embeddings into a lower-dimensional space, making clustering algorithms more effective. This process addresses the "curse of dimensionality," where increased dimensions can lead to sparse data and hinder clustering performance. 

**Common Dimensionality Reduction Techniques in BERTopic:**

1. **UMAP (Uniform Manifold Approximation and Projection):**
   - Captures both local and global structures of data, making it suitable for complex embeddings. 

2. **PCA (Principal Component Analysis):**
   - A linear method that projects data to a lower-dimensional space by maximizing variance.  [oai_citation_attribution:0‡en.wikipedia.org](https://en.wikipedia.org/wiki/Dimensionality_reduction)

3. **Truncated SVD (Singular Value Decomposition):**
   - Efficient for sparse data, commonly used in text mining. 

4. **cuML UMAP:**
   - A GPU-accelerated version of UMAP, beneficial for large datasets. 

5. **Skipping Dimensionality Reduction:**
   - Possible if embeddings are already in a manageable dimension. 

**Choosing the Right Technique:**

- **UMAP:** Ideal for complex, non-linear data structures where preserving both local and global relationships is essential.

- **PCA:** Suitable for linearly separable data and when interpretability of components is desired.

- **Truncated SVD:** Best for sparse datasets, such as term-document matrices in text mining.

- **cuML UMAP:** Optimal for large-scale data requiring faster processing.

- **Skipping Dimensionality Reduction:** Applicable when working with embeddings already in a low-dimensional space.

Selecting the appropriate dimensionality reduction technique enhances the performance and interpretability of topic modeling in BERTopic.

## UMAP
As a default, BERTopic uses UMAP to perform its dimensionality reduction. To use a UMAP model with custom parameters, we simply define it and pass it to BERTopic:

```python
from bertopic import BERTopic
from umap import UMAP

umap_model = UMAP(n_neighbors=15, n_components=5, min_dist=0.0, metric='cosine')
topic_model = BERTopic(umap_model=umap_model)
```

Here, we can define any parameters in UMAP to optimize for the best performance based on whatever validation metrics you are using.

## PCA

Although UMAP works quite well in BERTopic and is typically advised, you might want to be using PCA instead. It can be faster to train and perform inference. To use PCA, we can simply import it from sklearn and pass it to the umap_model parameter:

```python
from bertopic import BERTopic
from sklearn.decomposition import PCA

dim_model = PCA(n_components=5)
topic_model = BERTopic(umap_model=dim_model)
```

As a small note, PCA and k-Means have worked quite well in my experiments and might be interesting to use instead of PCA and HDBSCAN.

## Truncated SVD

Like PCA, there are a bunch more dimensionality reduction techniques in sklearn that you can be using. Here, we will demonstrate Truncated SVD but any model can be used as long as it has both a .fit() and .transform() method:

```python
from bertopic import BERTopic
from sklearn.decomposition import TruncatedSVD

dim_model = TruncatedSVD(n_components=5)
topic_model = BERTopic(umap_model=dim_model)
```

## cuML UMAP
Although the original UMAP implementation is an amazing technique, it may have difficulty handling large amounts of data. Instead, we can use cuML to speed up UMAP through GPU acceleration:

```python
from bertopic import BERTopic
from cuml.manifold import UMAP

umap_model = UMAP(n_components=5, n_neighbors=15, min_dist=0.0)
topic_model = BERTopic(umap_model=umap_model)
```

## Skip dimensionality reduction

Although BERTopic applies dimensionality reduction as a default in its pipeline, this is a step that you might want to skip. We generate an "empty" model that simply returns the data pass it to:

```python
from bertopic import BERTopic
from bertopic.dimensionality import BaseDimensionalityReduction

# Fit BERTopic without actually performing any dimensionality reduction
empty_dimensionality_model = BaseDimensionalityReduction()
topic_model = BERTopic(umap_model=empty_dimensionality_model)
```
